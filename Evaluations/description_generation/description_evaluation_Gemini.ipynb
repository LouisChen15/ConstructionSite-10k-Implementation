## SPICE

```python
score, scores = description_evaluator.get_spice_score(prediction_file, label_file)
print(sum(score)/len(score))
```

```python
print(scores[0])
```

```python
def result_assignment(arr, scor):
    arr[0] = scor['pr'] 
    arr[1] = scor['re'] 
    arr[2] = scor['f'] 
    
    return arr

all_arr = np.zeros((len(scores), 3))
rel_arr = np.zeros((len(scores), 3))
car_arr = np.zeros((len(scores), 3))
att_arr = np.zeros((len(scores), 3))
siz_arr = np.zeros((len(scores), 3))
col_arr = np.zeros((len(scores), 3))
obj_arr = np.zeros((len(scores), 3))

for i in range(len(scores)):
    all_arr[i] = result_assignment(all_arr[i], scores[i][0]["All"])
    rel_arr[i] = result_assignment(rel_arr[i], scores[i][0]["Relation"])
    car_arr[i] = result_assignment(car_arr[i], scores[i][0]["Cardinality"])
    att_arr[i] = result_assignment(att_arr[i], scores[i][0]["Attribute"])
    siz_arr[i] = result_assignment(siz_arr[i], scores[i][0]["Size"])
    col_arr[i] = result_assignment(col_arr[i], scores[i][0]["Color"])
    obj_arr[i] = result_assignment(obj_arr[i], scores[i][0]["Object"])

all_result = np.nanmean(all_arr, axis=0)
rel_result = np.nanmean(rel_arr, axis=0)
car_result = np.nanmean(car_arr, axis=0)
att_result = np.nanmean(att_arr, axis=0)
siz_result = np.nanmean(siz_arr, axis=0)
col_result = np.nanmean(col_arr, axis=0)
obj_result = np.nanmean(obj_arr, axis=0)

print(f"All: {all_result}")
print(f"Relation: {rel_result}")
print(f"Cardinality: {car_result}")
print(f"Attribute: {att_result}")
print(f"Size: {siz_result}")
print(f"Color: {col_result}")
print(f"Object: {obj_result}")
```
